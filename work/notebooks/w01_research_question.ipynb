## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os, subprocess
import pandas as pd

path = "data/raw/content_refresh_anonymized.csv"
if not os.path.exists(path):
    if not os.path.isdir("ml-internship-work"):
        subprocess.run(["git", "clone", "--depth", "1",
            "https://github.com/nimrafarooq2006-coder/ml-internship-work"], check=True)
    path = "ml-internship-work/" + path

df = pd.read_csv(path)
df["is_declining"] = df["trend_direction"].str.lower().eq("down").astype(int)
stale = df["days_since_last_update"] >= 180
visible = df["impressions_90d"] >= 500

print("Pages in the data:", len(df))
print("Share of pages declining:", round(df["is_declining"].mean(), 3))
print("Declining rate, stale pages (180+ days):", round(df.loc[stale, "is_declining"].mean(), 3))
print("Declining rate, other pages:", round(df.loc[~stale, "is_declining"].mean(), 3))
print("Pages that are stale AND visible:", int((stale & visible).sum()))
print("Number of stale pages:", int(stale.sum()))
```

Numbers from the starter data: 30,000 pages, and 54.2% are marked
declining. Only 174 pages (about 0.6%) are stale (180+ days since
update), and 17 of those are also visible (500+ impressions).
Stale pages were declining a bit less often (47.1%) than the other
pages (54.2%), but the stale group is small, so this difference may
be noise.

What this suggests: the simple rule "old page + lots of views" picks
out very few pages, so a hand rule alone may miss most declining
pages. This makes the lane worth studying, because a learned ranking
might find signals that a simple rule can't. These results are
directional only.
