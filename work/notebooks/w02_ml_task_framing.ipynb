## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import os, subprocess
import pandas as pd

path = "data/raw/content_refresh_anonymized.csv"
if not os.path.exists(path):
    if not os.path.isdir("ml-internship-work"):
        subprocess.run(["git", "clone", "--depth", "1",
            "https://github.com/nimrafarooq2006-coder/ml-internship-work"], check=True)
    path = "ml-internship-work/" + path

df = pd.read_csv(path)
df["is_declining"] = df["trend_direction"].str.lower().eq("down").astype(int)

unit = df[["content_id", "client_id", "impressions_90d",
           "days_since_last_update", "avg_position", "ctr",
           "word_count", "is_declining"]]

print("One row = one web page. Number of rows:", len(unit))
print(unit["is_declining"].value_counts())
unit.head(10)
```

One row is one web page. Each row has signals about the page
(impressions, days since update, position, CTR, word count) and my
target column is_declining (1 = declining, 0 = not). About 54% of the
pages are marked declining.

## 5. Why ML beats a fixed rule here

views" matched only 17 of 30,000 pages, so it misses most declining
pages. Many signals matter together (age, visibility, position, CTR),
and it is hard to pick one cutoff for each by hand. A model can
combine them and rank every page. It still has to prove itself: if it
doesn't beat the simple rule on Precision@50, I should keep the rule.
The output supports a real action: a reviewer checks the top-ranked
